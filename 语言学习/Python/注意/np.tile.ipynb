# np.tile的作用 —— 扩充矩阵行列
`np.tile(A, reps)` 的功能是将数组 A 按照 reps 指定的次数在各个维度上进行重复拼接，生成一个更大的数组。

```python
import numpy as np

arr = np.array([1, 2, 3])
print(np.tile(arr, 2))      # 在列方向重复2次
# 输出: [1 2 3 1 2 3]

print(np.tile(arr, (2, 1))) # 行重复2次，列重复1次（变成2行3列）
# 输出: 
# [[1 2 3]
#  [1 2 3]]

print(np.tile(arr, (2, 2))) # 行重复2次，列重复2次
# 输出:
# [[1 2 3 1 2 3]
#  [1 2 3 1 2 3]]
```
