# next(iter(...))
##
 核心功能
获取一个字典的第一个键（key）

## iter() 获取迭代器

- **作用**：将一个可迭代对象（如列表、元组、字符串、字典等）转换为迭代器。迭代器可以“记住”遍历的位置，每次取出一个元素。

```python
my_list = [10, 20, 30]
it = iter(my_list)
print(it)
```

## next() 从迭代器中取元素
- **作用**：每次调用 next()，迭代器会返回下一个元素。如果迭代器中没有元素了，会抛出 StopIteration 异常。
- **next(it, default)**：定义迭代完情况

```python
my_list = [10, 20, 30]
it = iter(my_list)

print(next(it))   # 输出 10
print(next(it))   # 输出 20
print(next(it))   # 输出 30

try:
    print(next(it))   # StopIteration 异常
except StopIteration as e:
    print(f'error:{e}')

print(next(it, None)) # 通过next(it, default)定义迭代完情况
```
