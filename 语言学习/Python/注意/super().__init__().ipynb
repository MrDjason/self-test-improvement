`super().__init__()`

`super().__init__()` 用于在子类的构造方法中调用父类的构造方法，以确保子类实例能够正确继承并初始化父类中定义的属性和逻辑。

```python
# 不使用 super().__init__() 的问题
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, breed):
        self.breed = breed

# 报错：Dog 没有 name 属性
d = Dog('柴犬')
print(d.breed)  # 输出: 柴犬
try:
    print(d.name)
except AttributeError as e:
    print(f"访问name出错：{e}") # AttributeError: 'Dog' object has no attribute 'name'
```

```python
# 使用 super().__init__()
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

# 报错：Dog 没有 name 属性
d = Dog('旺财', '柴犬')
print(d.breed) # 输出: 柴犬
print(d.name)  # 输出: 旺财
```
