# isinstance
- **说明**：isinstance 是 Python 内置函数，用于判断一个对象是否属于某个类型
- **基本语法**
```python
isinstance(object, type)
```

## 1.判断单个类型

````python
print(isinstance(3, int))
````

## 2.判断是否是多种类型中的一种（用元组）

````python
print(isinstance(3.4, (int, float)))
print(isinstance('abc', (int, str)))
````

## 3. 判断子类对象

````python
class MyDict(dict):
    pass

md = MyDict()
print(isinstance(md, (dict)))
print((type(md) is dict), type(md),sep='\n')
````
