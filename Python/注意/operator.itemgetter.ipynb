# operator.itemgetter

## 基本用法举例
### 1. 从元组/列表中按索引取元素

```python
import operator

get_second = operator.itemgetter(1) # 获取索引为1的元素

t = ('apple', 10, True)
print(get_second(t))
```

### 2. 同时取多个元素

```python
get_first_and_third = operator.itemgetter(0,2)
lst = [1, 200, 34, 7, 60]
print(get_first_and_third(lst))
```

### 3. 从字典中按键取值

```python
d = {'name':'david', 'age':18}
get_age = operator.itemgetter('age')
print(get_age(d))
```

### 4.sorted 中作为 key 函数

```python
students = [('Bob', 80), ('Alice', 95), ('Eve', 70)]
sorted_by_score = sorted(students, key=operator.itemgetter(1), reverse=True) # 对students中每一个元素都调用key=operator.itemgetter(1)进行排序，则对分数进行排序
print(sorted_by_score)
```
